### 示例4-1

```python
import requests
url="https://ecnugischaser.github.io/gis_develop_with_python/demo/html_encoding"
response = requests.get(url)
print(f"缺省的编码：{response.encoding}")
print("缺省编码显示的HTML文档：")
print(response.text)
print(f"分析出的编码：{response.apparent_encoding}")
response.encoding = response.apparent_encoding
print("按分析出编码显示的HTML文档：")
print(response.text)
```

### 示例4-4

```python
html_doc ="""
<html>
  <head>
    <title>HTML文档示例</title>
  </head>
  <body>
    <h1>华东师范大学</h1>
    <img id="a" src="photo1.jpg" >
    <div>
      <img id="b" src="photo2.jpg" >
    </div>
  </body>
</html>
"""
from bs4 import BeautifulSoup
soup = BeautifulSoup(html_doc,"lxml")
print("第一个img标签")
print(soup.img)
print("div标签中的第一个img标签")
print(soup.div.img)
print("所有img标签")
print(soup.find_all("img"))
print("id为a的img标签")
print(soup.find_all(id="a"))
```

### 示例4-5

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

user_agent = 'Mozilla/5.0 (Windows NT 10.0; WOW64; Trident/7.0; rv:11.0) like Gecko'
headers = {'User-Agent': user_agent}
url = 'https://baike.baidu.com/item/北京'
r = requests.get(url,headers=headers)
soup = BeautifulSoup(r.text,"lxml")
basic_info = soup.find(class_="basic-info J-basic-info cmn-clearfix")
dts = basic_info.find_all("dt")
item_name_list = []
for dt in dts:
    item_name = dt.get_text().strip("\n")
    item_name_list.append(item_name)

dds = basic_info.find_all("dd")
item_value_list = []
for dd in dds:
    item_value = dd.get_text().strip("\n")
    item_value_list.append(item_value)
    
frame = pd.DataFrame()
frame["item_name"] = item_name_list
frame["item_value"] = item_value_list
frame
```

### 示例4-6

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
driver = webdriver.Firefox()
driver.get('http://pythonscraping.com/pages/files/form.html'),
firstnameField = driver.find_element(By.NAME, "firstname")
lastnameField = driver.find_element(By.NAME, "lastname")
submitButton = driver.find_element(By.ID, "submit")

firstnameField.send_keys("小明")
lastnameField.send_keys("张"),
submitButton.click()
print(driver.find_element(By.TAG_NAME, "body").text)
driver.close()
```

### 示例4-7

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
driver = webdriver.Firefox()
driver.get('http://www.cnemc.cn/sssj/')
time.sleep(2)
a = driver.find_element(By.CSS_SELECTOR, "div#cityAirDailyScroll")
elements = a.find_elements(By.TAG_NAME, "li")
i = 0
for element in elements:
    i= i + 1
    element.click()
    cityAirDaily = element.text.split("\n")
    print(cityAirDaily)
    #这里只显示前 30 个城市
    if i >= 30:
        break
driver.close()
```
